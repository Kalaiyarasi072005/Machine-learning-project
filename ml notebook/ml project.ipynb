### Phase 4

```python

df_raw = df.copy()

print("Original dataset shape:", df_raw.shape)
print("Working dataset shape:", df.shape)
```

```python

df = df_raw.copy()

print(df.head())
```

```python

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

print(df["InvoiceDate"].dtype)
print(df["InvoiceDate"].head())
```

```python

print("Missing invoice dates:", df["InvoiceDate"].isna().sum())
```

```python

print(df.isnull().sum())
```

```python

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

print(missing_summary)
```

```python

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python

df = df.drop_duplicates().copy()

print("Shape after duplicate removal:", df.shape)
```

```python

df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print(
    df[["Quantity", "UnitPrice", "Revenue"]].head(10)
)
```

```python

df["Year"] = df["InvoiceDate"].dt.year

print(df[["InvoiceDate", "Year"]].head(10))
```

```python

df["Month"] = df["InvoiceDate"].dt.month

print(df[["InvoiceDate", "Month"]].head(10))
```

```python

df["MonthName"] = df["InvoiceDate"].dt.month_name()

print(df[["InvoiceDate", "Month", "MonthName"]].head(10))
```

```python

import numpy as np

conditions = [
    df["Quantity"] < 0,
    df["Quantity"] == 0,
    df["Quantity"] > 0
]

labels = [
    "Return_or_Cancellation",
    "Zero_Quantity",
    "Positive_Quantity"
]

df["TransactionType"] = np.select(
    conditions,
    labels,
    default="Unknown"
)

print(df["TransactionType"].value_counts())
```

```python

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nFirst 5 transformed records:")
print(df.head())
```

```python

from pathlib import Path

output_dir = Path("output")
output_dir.mkdir(parents=True, exist_ok=True)

df.to_csv(
    output_dir / "transformed_ecommerce_data.csv",
    index=False
)

print("Transformed dataset saved successfully!")
```

### Phase 5


```python

total_revenue = df["Revenue"].sum()

print("Total Revenue:", round(total_revenue, 2))
```

```python

total_quantity = df["Quantity"].sum()

print("Net Quantity:", total_quantity)
```

```python

country_revenue = (
    df.groupby("Country")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

print(country_revenue)
```

```python

top_5_countries = country_revenue.head(5)

print(top_5_countries)
```

```python

product_revenue = (
    df.groupby("Description")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

print(product_revenue.head(10))
```

```python

product_quantity = (
    df.groupby("Description")["Quantity"]
      .sum()
      .sort_values(ascending=False)
)

print(product_quantity.head(10))
```

```python

monthly_revenue = (
    df.groupby(["Year", "Month"])["Revenue"]
      .sum()
      .reset_index()
      .sort_values(["Year", "Month"])
)

print(monthly_revenue)
```

```python

monthly_summary = (
    df.groupby(["Year", "Month"])
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Transaction_Lines=("InvoiceNo", "size"),
          Unique_Invoices=("InvoiceNo", "nunique")
      )
      .reset_index()
      .sort_values(["Year", "Month"])
)

print(monthly_summary)
```

```python

customer_summary = (
    df.dropna(subset=["CustomerID"])
      .groupby("CustomerID")
      .agg(
          Total_Spending=("Revenue", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Invoice_Count=("InvoiceNo", "nunique")
      )
      .sort_values("Total_Spending", ascending=False)
)

print(customer_summary.head(10))
```

```python

kpi_summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Unique Invoices",
        "Unique Products",
        "Unique Countries",
        "Known Customers",
        "Total Revenue",
        "Net Quantity"
    ],
    "Value": [
        len(df),
        df["InvoiceNo"].nunique(),
        df["StockCode"].nunique(),
        df["Country"].nunique(),
        df["CustomerID"].nunique(),
        round(df["Revenue"].sum(), 2),
        df["Quantity"].sum()
    ]
})

print(kpi_summary)
```

```python

from pathlib import Path

output_dir = Path("output")
output_dir.mkdir(parents=True, exist_ok=True)

country_revenue.rename(
    "Revenue"
).to_csv(output_dir / "country_revenue.csv")

product_revenue.rename(
    "Revenue"
).to_csv(output_dir / "product_revenue.csv")

monthly_summary.to_csv(
    output_dir / "monthly_summary.csv",
    index=False
)

kpi_summary.to_csv(
    output_dir / "kpi_summary.csv",
    index=False
)

print("Phase 05 reports saved successfully!")
```
